# Part 1: Derived Magnetometer Data

The CSA Open Data Portal gives raw time-series measurements for CARISMA data (X, Y, Z, in nT). The CARISMA team also published a set of higher-level derived products that are computed from the raw data and are hosted directly at [data.carisma.ca](https://data.carisma.ca)

This is the same instrument network, but with the signal processing already complete. 

---

The archive is split into two top level folders, one per instrument type. The FGM folder includes the Fluxgate Magnetometer data, and the ICM folder includes the Induction Coil Magnetometer data. 

```
https://data.carisma.ca/
|
|-- FGM/                                 <- Fluxgate Magnetometer data
|   |-- 1Hz/                             <- Raw data at 1Hz (CSV)
|   |-- 8Hz/                             <- Raw data at 8Hz (CSV)
|   |-- CDF1Hz/                          <- Same data in CDF format
|   |-- CDF8Hz/
|   |-- PC1PowerSpectra/                 <- Derived: Pc1 wave power spectra
|   |-- PC5PowerSpectra/                 <- Derived: Pc5 wave power spectra
|   |-- PlasmaDensity/                   <- Derived: Magnetospheric plasma density
|   |-- CAEIndex/                        <- Derived: Canadian auroral electrojet index
|
|-- ICM/                                 <- Induction Coil Magnetometer data
|   |-- 20Hz/                            <- Raw data at 20Hz (CSV)
|   |-- 100Hz/                           <- Raw data at 100Hz (CSV)
|   |-- CDF20Hz/                         <- Same data in CDF format
|   |-- CDF100Hz/
|   |-- PowerSpectra/                    <- Derived: High Frequency Power Spectra
|   |-- EllipticityAndPolarisation/      <- Derived: Wave Polarization Analysis
|   |-- Audio                            <- Derived: Audio band Wave files
```

**FGM** (Fluxgate Magnetometer) measures the absolute magnetic field vector: the same X, Y, Z from Tutorial 01: CARISMA Magnetometer Data. It is best for slowly-varying signals (mHz frequency range).

**ICM** (Induction Coil Magnetometer) measures the rate of change of the magnetic field (dB/dT), making it more sensitive to rapid ascillations in the sub-Hz to 100Hz range.

## 1.2 Power Spectra

**What is it?** The raw magnetometer time series is transformed from the time domain into the frequency domain using the Fast Fourier Transform (FFT). The results is a power spectrum, a plot of how much the wave energy exists at each frequnecy for a given time window. 

CARISMA produces spectra for two physically important Ultra Low Frequency (ULF) wave bands:

| Product | Location | Frequnecy Band | Physical Source |
|---------|----------|----------------|-----------------|
| **PC5 Spectra** | `FGM/PC5PowerSpectra/`| ~1-7mHz | Solar wind pressure pulses, field line resonsances, driven by the Kelvin-Helmholtz instability on the magnetopause |
| **PC1 Spectra** | `FGM/PC1PowerSpectra/`| ~0.2-5mHz | Electromagnetic Ion Cyclotron (EMIC) waves generated in the inner magnetosphere |
| **ICM Spectra** | `ICM/PowerSpectra/`| Up to 100mHz | Higher frequnecy waves also captures Pc1 with much better sensitivity than FGM |

**Why do they matter?** ULF waves are responsible for accelerating and scattering radiation belt electrons. Researches use these spectra to track wave activity without having to process hours of raw time series themselves. 

## 1.3 Ellipticity and Polarization
**What are they?** A magnetic wave doesn't just oscillate in one direction, but the field vector can rotate as the wave passes. This rotation can be characterized with two horizontal components (Bx and By, aka North and East).

- **Linear Polarization**: the field oscillates back and forth along a fixed acis (like a pendulum)
- **Cirular Polarization**: the field vector traces a circle over each wave cycle
- **Elliptical Polarization**: the general case, where the field traces an ellipse

**Ellipticity** is a number from -1 to +1 describing:
- `-1` = left-hand circular
- `0` = purely linear
- `1` = right-hand circular

**Why does it matter?** Different wave types have characteristic polarization signatures. EMIC waves are left-hand polarized (negative ellipticity), which is physically signficant because the handedness matches the gyration of ions in the magnetic field (so the waves interact strongly with ring current protons). Identifying polarization is a key step in classifying waves.

These products are computed from the ICM data (which captures the relevant frequency range) and are found at:

```
https://data.carisma.ca/ICM/EllipticityAndPolarisation/
```

## 1.4 Plasma Density

**What is it?** This is an estimate of the plasma mass density in the magnetosphere, derived from the ground0based magnetometer data

The magnetosphere supports standing Alfvén waves along magnetic field lines called the Field Line Resonances (FLRs). The frequnecy at which a given field line resonsates depends on its length and the density of plasma sitting on it, similar to a guitar string's length and tension. 

By identifying the resonance frequnecies in the Pc5 power spectra, CARISMA can invert the relationship to estimate plasma density several Earth radii above the ground:

```
Resonance frequency -> Field line length (known from models) -> Plasma density
```

This is found at 
```
https://data.carisma.ca/FGM/PlasmaDensity/
```

It provides continuous ground-based coverage of the plasmosphere and inner magnetosphere.
