# Partie 1 : Données magnétométriques dérivées

Le portail de données ouvertes de l'ASC fournit les mesures brutes en série temporelle pour les données CARISMA (X, Y, Z, en nT). L'équipe CARISMA publie également un ensemble de produits dérivés de haut niveau calculés à partir des données brutes et hébergés directement sur [data.carisma.ca](https://data.carisma.ca)

Il s'agit du même réseau d'instruments, mais avec le traitement du signal déjà effectué.

---

L'archive est divisée en deux dossiers principaux, un par type d'instrument. Le dossier FGM contient les données du fluxgate magnetomètre, et le dossier ICM contient les données de l'induction coil magnetomètre.

```
https://data.carisma.ca/
|
|-- FGM/                                 <- Données Fluxgate Magnetometer
|   |-- 1Hz/                             <- Données brutes à 1Hz (CSV)
|   |-- 8Hz/                             <- Données brutes à 8Hz (CSV)
|   |-- CDF1Hz/                          <- Même données au format CDF
|   |-- CDF8Hz/
|   |-- PC1PowerSpectra/                 <- Dérivé : spectres de puissance Pc1
|   |-- PC5PowerSpectra/                 <- Dérivé : spectres de puissance Pc5
|   |-- PlasmaDensity/                   <- Dérivé : densité plasmatique
|   |-- CAEIndex/                        <- Dérivé : indice électrojet auroral canadien
|
|-- ICM/                                 <- Données Induction Coil Magnetometer
|   |-- 20Hz/                            <- Données brutes à 20Hz (CSV)
|   |-- 100Hz/                           <- Données brutes à 100Hz (CSV)
|   |-- CDF20Hz/                         <- Même données au format CDF
|   |-- CDF100Hz/
|   |-- PowerSpectra/                    <- Dérivé : spectres de puissance haute fréquence
|   |-- EllipticityAndPolarisation/      <- Dérivé : polarisation et ellipticité
|   |-- Audio                            <- Dérivé : fichiers audio (bandes ULF)
```

**FGM** (Fluxgate Magnetometer) mesure le vecteur de champ magnétique absolu : les mêmes X, Y, Z que dans le Tutoriel 01 : CARISMA Magnetometer Data. Il est adapté aux signaux à variation lente (gamme de fréquences mHz).

**ICM** (Induction Coil Magnetometer) mesure la dérivée temporelle du champ magnétique (dB/dT), ce qui le rend plus sensible aux oscillations rapides dans la gamme sub-Hz à 100Hz.

## 1.2 Spectres de puissance

**Qu'est-ce que c'est ?** La série temporelle magnétométrique brute est transformée du domaine temporel au domaine fréquentiel à l'aide de la Transformée de Fourier Rapide (FFT). Le résultat est un spectre de puissance, un graphique montrant l'énergie d'onde présente à chaque fréquence pour une fenêtre temporelle donnée.

CARISMA produit des spectres pour deux bandes Ultra Low Frequency (ULF) physiquement importantes:

| Produit | Emplacement | Bande de fréquences | Source physique |
|---------|-------------|----------------------|------------------|
| **Spectres PC5** | `FGM/PC5PowerSpectra/`| ~1-7 mHz | Pulsations de pression du vent solaire, résonances de lignes de champ, entraînées par l'instabilité de Kelvin-Helmholtz sur la magnétopause |
| **Spectres PC1** | `FGM/PC1PowerSpectra/`| ~0.2-5 mHz | Ondes Ionéatoires Cyclotroniques Électromagnétiques (EMIC) générées dans la magnétosphère interne |
| **Spectres ICM** | `ICM/PowerSpectra/`| Jusqu'à 100 mHz | Ondes de plus haute fréquence capturées avec une sensibilité supérieure par rapport au FGM |

**Pourquoi cela importe-t-il ?** Les ondes ULF sont responsables de l'accélération et de la diffusion des électrons de la ceinture de radiation. Les chercheurs utilisent ces spectres pour suivre l'activité d'ondes sans avoir à traiter eux-mêmes des heures de séries temporelles brutes.

## 1.3 Ellipticité et Polarisation
**Qu'est-ce que c'est ?** Une onde magnétique n'oscille pas seulement dans une direction : le vecteur de champ peut tourner pendant le passage de l'onde. Cette rotation peut être caractérisée à l'aide des deux composantes horizontales (Bx et By, soit Nord et Est).

- **Polarisation linéaire** : le champ oscille d'avant en arrière le long d'un axe fixe (comme un pendule)
- **Polarisation circulaire** : le vecteur de champ trace un cercle à chaque cycle d'onde
- **Polarisation elliptique** : cas général, où le vecteur trace une ellipse

**L'ellipticité** est un nombre de -1 à +1 décrivant :
- `-1` = circulaire gauche
- `0` = purement linéaire
- `1` = circulaire droite

**Pourquoi est-ce important ?** Différents types d'ondes présentent des signatures de polarisation caractéristiques. Les ondes EMIC sont polarisées à gauche (ellipticité négative), ce qui est physiquement significatif car le sens de rotation correspond à la gyration des ions dans le champ magnétique (les ondes interagissent donc fortement avec les protons du courant annulaire). Identifier la polarisation est une étape clé pour classer les ondes.

Ces produits sont calculés à partir des données ICM (qui couvrent la gamme de fréquences pertinente) et se trouvent à :

```
https://data.carisma.ca/ICM/EllipticityAndPolarisation/
```

## 1.4 Densité plasmatique

**Qu'est-ce que c'est ?** Il s'agit d'une estimation de la masse volumique du plasma dans la magnétosphère, dérivée des magnétomètres au sol.

La magnétosphère supporte des ondes d'Alfvén stationnaires le long des lignes de champ magnétique appelées résonances de lignes de champ (Field Line Resonances, FLR). La fréquence à laquelle une ligne de champ donnée résonne dépend de sa longueur et de la densité de plasma qui s'y trouve, similaire à la façon dont la longueur et la tension d'une corde de guitare déterminent sa fréquence.

En identifiant les fréquences de résonance dans les spectres de puissance Pc5, CARISMA peut inverser la relation pour estimer la densité plasmatique à plusieurs rayons terrestres au-dessus du sol :

```
Fréquence de résonance -> Longueur de la ligne de champ (connue à partir de modèles) -> Densité plasmatique
```

Cela se trouve à
```
https://data.carisma.ca/FGM/PlasmaDensity/
```

Il fournit une couverture continue basée au sol de la plasmasphère et de la magnétosphère interne.
